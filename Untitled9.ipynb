Variable aleatoria binomial negativa


```python
import numpy as np
import scipy.stats as sps
import matplotlib.pyplot as plt

r = 10

p = 0.4

s = 100000

np.random.seed(3)


nbinom_fracasos = sps.nbinom.rvs(r, p, size=s)
nbinom_ensayos = nbinom_fracasos + r

plt.figure(figsize=(10, 6))
plt.hist(
    nbinom_ensayos,
    density=True,
    bins=np.arange(r, np,max(nbinom_ensayos) +1) - 0.5,
    color="skyblue",
    edgecolor="grey",
    label='Simulación (Ensayos totales)'
)

x_teorico = np.arange(r, np.max(nbinom_ensayos))
pmf_teorica = sps.nbinom.pmf(x_teorico, r, p)

```
