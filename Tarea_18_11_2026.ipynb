```python
# ==========================================
# EJERCICIO 1: Una ficha numerada
# ==========================================

# 1. Construcción del espacio muestral y eventos con comprensiones de conjuntos
Omega_1 = set(range(1, 13))
A_1 = {n for n in Omega_1 if n % 2 == 0}
B_1 = {n for n in Omega_1 if n > 8}

print("--- Ejercicio 1 ---")
print(f"Espacio muestral Omega_1: {Omega_1}")
print(f"Evento A_1 (pares): {A_1}")
print(f"Evento B_1 (> 8): {B_1}")

# 2. Elementos de la intersección y verificación
interseccion_1 = A_1 & B_1
print(f"A_1 ∩ B_1 (verificación con &): {interseccion_1}")

# 3. Cálculo de probabilidades condicionales
p_a_1 = len(A_1) / len(Omega_1)
p_b_1 = len(B_1) / len(Omega_1)
p_a_given_b_1 = len(interseccion_1) / len(B_1)
p_b_given_a_1 = len(interseccion_1) / len(A_1)

print(f"P(A) = {p_a_1}")
print(f"P(B) = {p_b_1}")
print(f"P(A | B) = {p_a_given_b_1}")
print(f"P(B | A) = {p_b_given_a_1}")

# 4. Verificación de independencia
independientes_1 = (p_a_given_b_1 == p_a_1)
print(f"¿Son A y B independientes? {independientes_1} (P(A|B) = P(A))")

```

```python

# ==========================================
# EJERCICIO 2: Dos fichas sin reemplazo
# ==========================================

# 1. Construcción del espacio muestral (excluyendo pares idénticos i == j)
Omega_2 = {(i, j) for i in range(1, 6) for j in range(1, 6) if i != j}

print("\n--- Ejercicio 2 ---")
print(f"Total de elementos en Omega_2 (sin reemplazo): {len(Omega_2)}")

# 2. Definición de A2, B2 y su intersección
A_2 = {(i, j) for i, j in Omega_2 if i % 2 == 0}
B_2 = {(i, j) for i, j in Omega_2 if j % 2 == 0}
interseccion_2 = A_2 & B_2

print(f"A_2 (primera ficha par): {A_2}")
print(f"B_2 (segunda ficha par): {B_2}")
print(f"A_2 ∩ B_2 (verificación con &): {interseccion_2}")

# 3. Cálculo de probabilidades
p_b_2 = len(B_2) / len(Omega_2)
p_b_given_a_2 = len(interseccion_2) / len(A_2)
p_a_given_b_2 = len(interseccion_2) / len(B_2)

print(f"P(B_2) = {len(B_2)} / {len(Omega_2)} = {p_b_2}")
print(f"P(B_2 | A_2) = {len(interseccion_2)} / {len(A_2)} = {p_b_given_a_2}")
print(f"P(A_2 | B_2) = {len(interseccion_2)} / {len(B_2)} = {p_a_given_b_2}")

# 4. Verificación de independencia
independientes_2 = (p_b_given_a_2 == p_b_2)
print(f"¿Son A_2 y B_2 independientes? {independientes_2} (P(B_2|A_2) != P(B_2))")
```
